```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name : str = ""
    age : int
```

```python
user = User(name="Hita", age=28)
user
```

#### Field Validation

```python
class Product(BaseModel):
    title:str = Field(max_length=15)
    des:str = Field(min_length=20)
    price:int = Field(gt=0, le=100)
```

```python
product = Product(title="Samsung S11", des="Latest Model from Samsung", price=90)
product2 = Product(title="Samsung S15", des="Latest Model from Samsung", price=90)
```

```python
class Order(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    product:Product
```

```python
order = Order(id = "san111", product=product)
```

```python
order.product.title
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items: List[Product]
    comment: str = Field(default="NA")
```

```python
# cart = Carts(id = "trkl12", items=[product])
cart = Carts(id = "trkl12", items=[product, product2], comment = "My Cart")
```

```python
cart
```

```python
for product in cart.items:
    print(product.title)
```

```python
# Annotated

# commonValidator = Annotated[List[Product], "This is Common Validator"]
commonValidator = Annotated[List[Product], Field(default=[])]
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items: commonValidator
    comment: str = Field(default="NA")
```

```python
class Order(BaseModel):
    products: commonValidator
```

```python
# ord = Order()
ord = Order(products=[product, product2])
```
