#### Document Text Loader

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("../data/bhakt_prahlad_story.txt", encoding="utf-8")
texts = loader.load()
```

```python
pdfloader = PyPDFLoader("../data/KedarkanthaKotgaonItinerary.pdf")
pdfs = pdfloader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 100, chunk_overlap = 30)
```

```python
text_chunks = splitter.split_documents(texts)
len(text_chunks)
```

```python
pdf_chunks = splitter.split_documents(pdfs)
len(pdf_chunks)
```
